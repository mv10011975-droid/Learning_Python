Range of projectile depend on its initial speed and angle 
$$Range = \frac{v_{\theta}^2 sin(2\theta)}{g}$$
Similar to range maximum height of any projectile only depend on its initial speed and angle 
$$Max\ height=\frac{v_{\theta}^2 sin^2(\theta)}{2g}$$
Position of protectile at any time is depend on its initial speed, angle and time
$$x=v_{\theta} cos(\theta) t$$
$$y=v_{\theta} sin(\theta) t-\frac{1}{2}g t^2$$

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

In our formula we have 4 varibles which are :
- Initial velocity $v_{0}$
- Angle from ground $\theta$
- Acceleration due to gravity $g$
- Time $t$

```python
while True:
    try:
        angle=float(input("Write angle of projrctile from ground in degree:"))
        v=float(input("Write initial velocity of projectile in meter per second:"))
        a=angle*np.pi/180
        break
    except ValueError: 
        print("Values should be numerical only")

g=9.8
mxh=v**2*np.sin(a)**2/(2*g)
rg=v**2*np.sin(2*a)/g
print(f"Range of projectile is:{rg:.2f}m")
print(f"Maximum height of projectile is:{mxh:.2f}m")
while True:
    while True:
        try:
            print("Do you want to check position of x and y at any specfic time(yes/no)")
            u=input()
            if u.lower() not in ["yes","no"]:
                raise ValueError
            break
        except ValueError:
            print("Answer only in yes or no")
    tf=2*v*np.sin(a)/g
    if u.lower()=="yes":
        while True:
            try:
                t=float(input("Write time in seconds:"))
                if t<0:
                    raise ValueError
                break
            except ValueError:
                print("Give appropriate value for t")
        if t<tf:
            x=v*np.cos(a)*t
            y=v*np.sin(a)*t-(g*t**2)/2
            print(f"X coordinate of particle at time{t}is:{x:.2f}")
            print(f"Y coordinate of particle at time{t}is:{y:.2f}")
        else:
            x=v*np.cos(a)*tf
            y=0
            print(f"X coordinate of particle at time{t}is:{x:.2f}")
            print(f"Y coordinate of particle at time{t}is:{y:.2f}")
    elif u.lower()=="no":
        print("Thankyou for visiting")
        break
        
```

## Ploting the trajectory

```python
%matplotlib inline
t=np.linspace(0,tf,100000)
x=x=v*np.cos(a)*t
y=v*np.sin(a)*t-(g*t**2)/2
plt.figure(figsize=(10,5))
plt.plot(x,y , linewidth=2)
plt.xlabel("Horizontal distance(m)")
plt.ylabel("Verticle height(m)")
plt.title("Position of particle")
plt.show()
```

### Observation
- After ploting graph we can see it follows path of a **parabola**
- when we replace **t** in both **x** and **y** we will get a quadratic equation as $$y=x tan(\theta)-\frac{1}{2 v^2} g x^2 sec(\theta)^2$$
- As we can see in equation there is **-** with $x^2$ so it is equation of downward fasing parabola which is what happening in graph
