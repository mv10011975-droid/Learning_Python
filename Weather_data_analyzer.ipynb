# Objective
Analyze real weather data using pandas, Numpy and Matplotlib.  
We will find pattern in temprature, humidity, and rainfall across different months

## Tools used
- `pandas` — loading and analyzing data
- `Numpy` - Making list of data
- `Matplotlib` - Plot graphs

## Dataset
We use a synthetic but realistic weather dataset with:
- Monthly temperature (°C)
- Humidity (%)
- Rainfall (mm)
- Wind speed (km/h)

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Creating the dataset
we will create a realistic one-year weather dataset for analysis


```python
data={
    "Month":["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Temprature":[12,13,20,29,37,39,25,24,22,19,13,11],
    "Humidity":[65, 60, 55, 50, 45, 60, 80, 85, 75, 65, 65, 68],
    "Rainfall":[15, 12, 8, 5, 10, 80, 180, 160, 90, 30, 15, 12],
    "Windspeed":[12, 14, 16, 18, 20, 22, 18, 16, 15, 13, 12, 11]
}
df= pd.DataFrame(data)
print(df)
```

## Basic statics
let us calculate key statics for each weather variable

```python
print("_____Weather statistics_____")
print(df.describe().round(2))

print("_____Temprature details_____")
temp=df["Temprature"].values
print(f"Hottest month:{df["Month"][np.argmax(temp)]}({np.max(temp)}C)")
print(f"Coldest month:{df["Month"][np.argmin(temp)]}({np.max(temp)}C)")
print(f"Average temprature:{df["Temprature"].mean():.2f}C")
print(f"Temprature standard deviation:{df["Temprature"].std():.2f}C")

print("_____Rainfall Deatils_____")
rain=df["Rainfall"].values
print(f"Wettest month: {df['Month'][np.argmax(rain)]} ({np.max(rain)}mm)")
print(f"Drtest month: {df["Month"][np.argmin(rain)]}({np.min(rain)}mm)")
print(f"Average Rainfall:{df["Rainfall"].mean():.2f}mm")
print(f"total Rainfall:{df["Rainfall"].sum()}mm")
```

## Visualizations

### Graph 1 — Monthly Temperature

```python
plt.figure(figsize=(10,4))
plt.plot(
    df["Month"],
    df["Temprature"],
    marker="o",
    linestyle="--")
plt.xlabel("Months")
plt.ylabel("Temprature\nC")
plt.title("Temprature vs Month")

plt.axhline(y=df["Temprature"].mean(),label=f"Tem avg:{df["Temprature"].mean():.2f}",color="red")
plt.legend()
plt.show()
```

### Graph 2 — Monthly Rainfall

```python
plt.figure(figsize=(10,4))
plt.plot(df["Month"],df["Rainfall"],marker="o",markersize=6,color="coral")
plt.xlabel("Months")
plt.ylabel("Rainfall\nmm")
plt.axhline(
    y=df["Rainfall"].mean(),
    label=f"Avg rainfall:{df["Rainfall"].mean():.2f}mm",
    color="coral")
plt.legend()
plt.show()
```

### Graph 3 — Temperature vs Humidity

```python
plt.figure(figsize=(10,4))
plt.plot(
    df["Month"],
    df["Temprature"],
    marker="o",
    linestyle="--")
plt.xlabel("Months")
plt.ylabel("Temprature(C) / Humidity(%)")
plt.title("Temprature vs Humidity")

plt.plot(
    df["Month"],
    df["Humidity"],
    marker="o",
    linestyle="--")
plt.legend()
plt.show()
```
